M = Marks → add

Expected result: 95
Start with score = 0
M:50 → Add 50 → score = 50
B:10 → Add 10 → score = 60
M:30 → Add 30 → score = 90
P:5 → Subtract 5 → score = 85
B:10 → Add 10 → score = 95

```python
def Stud_M(l):
    score = 0

    for marks in l:
        type,items = marks.split(":")
        marks = int(items)

        if type == "M":
            score += marks
        elif type == "B":
            score += marks
        else :
             score -= marks


    return score
print(Stud_M(["M:50", "B:10", "M:30", "P:5", "B:10"]))
```

4. Employee Salary

Write a function that accepts salary transactions.

S = Basic salary
B = Bonus
T = Tax
D = Deduction

Example:

["S:30000", "B:5000", "T:3000", "D:1000"]

Return the final salary.

```python
def Employee_Sal(K):
    S = B = T = D = 0

    for s1 in K:
        s2 = s1.split(":")

        if s2[0] == "S":
            S += int(s2[1])
        elif s2[0] == "B":
            B += int(s2[1])
        elif s2[0] == "T":
            T -= int(s2[1])
        elif s2[0] == "D":
            D -= int(s2[1])

    return S + B + T + D


print(Employee_Sal(["S:30000", "B:5000", "T:3000", "D:1000"]))
```

```python
def Employee_Sal(K):
    S = B = T = D = 0

    for s1 in K:
        s2 = s1.split(":")

        if s2[0] == "S":
            S += int(s2[1])
        elif s2[0] == "B":
            B += int(s2[1])
        elif s2[0] == "T":
            T += int(s2[1])
        elif s2[0] == "D":
            D += int(s2[1])

    return S + B - T - D


print(Employee_Sal(["S:30000", "B:5000", "T:3000", "D:1000"]))
```
