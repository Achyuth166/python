```python
#1.Find the first non-repeated character in a string.
s = input()

for i in s:
    if s.count(i) == 1:
        print(i)
        break
```

```python
#2.Reverse each word in a sentence.
s = input()

words = s.split()

for word in words:
    rev = ""
    
    for ch in word:
        rev = ch + rev
    
    print(rev, end=" ")
```

```python
#3.Find the longest word in a sentence.

s = input()

words = s.split()
l = []

for v in words:
    l.append((len(v), v))

l.sort()

print(l[-1][1])
```

```python
#4.Count vowels in a string without using count().

s = input()

cnt = 0

for ch in s:
    if ch in "aeiouAEIOU":
        cnt += 1

print(cnt)
```

```python
#5.Check whether a string is a palindrome without using slicing#
s = input()
rev = ""

for var in (s):
    rev = var + rev

if s == rev:
    print("palindrone")
else:
    print("not palindrone")
```
