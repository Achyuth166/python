### big of 2

```python
num1 = int(input("enetr no1"))
num2 = int(input("enter no2"))

if (num1 > num2):
    print("Num 1 is big",num1)
else:
    print("Num 2 is big",num2)
```

```python
large = 0

if (num1 > num2):
    large =  num1
else:
    large = num2

print("Biggest num is",large)
```

### even or odd

```python
num = int(input("enetr number for even or odd"))

res = ''

if ( num % 2 == 0 ):
    res += "even"
else:
    res += "odd"

print("The resulr is ",res)
    
```

### Num is Pos / Neg / Zero

```python
num = eval(input("eneter a number"))

res = ''
if (num > 0):
    res = "Positive"
elif (num < 0):
    res = "Negative"
else:
    res = "Zero"
    
print(res)
```
