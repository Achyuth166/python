```python
def count_substring(string,substring):
    count = 0

    i = string.find(substring)
    while i != -1:
        count += 1
        i = string.find(substring,i+1)
    return count

print(count_substring("CDC"))
```

```python
def count_num(l):
    D = W = 0

    for str1 in l:
        str2 = str1.split(":")

        if str2[0] == "D":
            D = D + int(str2[1])
        else:
            W = W + int(str2[1])

    return D, W


print(count_num(["D:5000", "W:2000", "D:3000", "W:1000"]))
```

```python
def count_num(l):
    d = w = 0

    for item in l:
        parts = item.split(":")
        amount = int(parts[1])

        if parts[0] == "D":
            d += amount
        elif parts[0] == "W":
            w += amount

    return d, w


print(count_num(["D:5000", "W:2000", "D:3000", "W:1000"]))
```
