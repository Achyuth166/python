```python
def multi_num(n):
    for i in range(1, 11):
        print(f"{n} * {i} = {n * i}")


num = int(input("Enter the number: "))
multi_num(num)
```

```python
def fact(n):
    f = 1
    for i in range(1, n + 1):
        f *= i
    return f
n = int(input("Enter the number: "))
result = fact(n)
print(result)
```

```python
def even(n):
    count = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            count += 1

    return count


n = int(input("Enter the number: "))

result = even(n)

print(result)
```
