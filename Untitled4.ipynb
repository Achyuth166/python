```python
num1 = int(input("Enter total number of classes conducted: "))
num2 = int(input("Enter number of classes attended: "))

attendance = (num2 / num1) * 100

print("Attendance percentage =", attendance)

if attendance >= 75:
    print("Eligible to write the final examination")
else:
    print("Not eligible to write the final examination")
```
