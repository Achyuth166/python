```python
#1.Find the first non-repeated character in a string.
s = input()

for i in s:
    if s.count(i) == 1:
        print(i)
        break
```

```python
#2.Reverse each word in a sentence.
s = input()

words = s.split()

for word in words:
    rev = ""
    
    for ch in word:
        rev = ch + rev
    
    print(rev, end=" ")
```

```python
#3.Find the longest word in a sentence.

s = input()

words = s.split()
l = []

for v in words:
    l.append((len(v), v))

l.sort()

print(l[-1][1])
```

```python
#4.Count vowels in a string without using count().

s = input()

cnt = 0

for ch in s:
    if ch in "aeiouAEIOU":
        cnt += 1

print(cnt)
```

```python
#5.Check whether a string is a palindrome without using slicing#
s = input()
rev = ""

for var in (s):
    rev = var + rev

if s == rev:
    print("palindrone")
else:
    print("not palindrone")
```

```python
def sum_prod(list):
    sum,prod = 0,1

for var in list:
    sum = sum + var
    prod = prod * var
return sum , prod

    
list = [ 1,2,44,5,66,7]
print(sum_prod(list))
    
```

```python
def max_num(list):
    max_num = list[0]

    for var in list:
        if(var > max_num):
            max_num = var
        else:
            continue


    return max_num

res = max_num([2,17,2,1,34,45,6])
print(res)
```
