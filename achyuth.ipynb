reverse of number while loop 

```python
num = int(input("entre the number"))

reverse_number = 0

while num != 0:
    digit = num % 10
    reverse_number = reverse_number * 10 + digit
    num = num // 10

print("reverse_number:", reverse_number)
```

reverse of number using for loop

```python
num = int(input("Enter the number: "))

reverse_number = 0

for i in range(len(str(num))):
    digit = num % 10
    reverse_number = reverse_number * 10 + digit
    num = num // 10

print("Reverse number:", reverse_number)
```

perfect number loop 

```python
num = int(input("Enter the number: "))

sum = 0
i = 1

while i <= num // 2:
    if num % i == 0:
        sum = sum + i
    i += 1

if sum == num:
    print(num, "perfect number")
else:
    print(num, "not perfect number")
```

```python
num = int(input("Enter a number: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count = count + 1

if count == 2:
    print("Prime Number")
else:
    print("Not Prime Number")
```

 print 3 Largest number
 

```python
n = int(input("How many numbers do you want to enter: "))

largest1 = float("-inf")
largest2 = float("-inf")
largest3 = float("-inf")

i = 0

while i < n:
    num = int(input("Enter the number: "))

    if num > largest1:
        largest3 = largest2
        largest2 = largest1
        largest1 = num

    elif num > largest2:
        largest3 = largest2
        largest2 = num

    elif num > largest3:
        largest3 = num

    i += 1

print("Largest number:", largest1)
print("Second largest:", largest2)
print("Third largest:", largest3)
```

3 largest numbers using for loop

```python
n = int(input("How many numbers do you want to enter: "))

largest1 = float("-inf")
largest2 = float("-inf")
largest3 = float("-inf")

for i in range(n):
    num = int(input("Enter the number: "))

    if (num > largest1):
        largest3 = largest2
        largest2 = largest1
        largest1 = num

    elif num > largest2:
        largest3 = largest2
        largest2 = num

    elif num > largest3:
        largest3 = num

print("Largest number:", largest1)
print("Second largest:", largest2)
print("Third largest:", largest3)
```

factorial number

```python
def fact(n):
    f = 1
    for i in range(1, n + 1):
        f *= i
    return f
n = int(input("Enter the number: "))
result = fact(n)
print(result)
```

Count Even Numbers Create a function that accepts n and counts how many even numbers exist between 1 and n.

```python
def even(n):
    count = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            count += 1

    return count


n = int(input("Enter the number: "))

result = even(n)

print(result)
```
